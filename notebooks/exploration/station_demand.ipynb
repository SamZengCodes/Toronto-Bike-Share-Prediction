```python
print(pd.read_sql("SELECT Start_Station_Id, Start_Station_Name, strftime('%Y-%m-%d-%H:00', Start_Time), AVG(Trip_Duration), COUNT(*) FROM trips GROUP BY Start_Station_id, strftime('%Y-%m-%d-%H', Start_Time) ORDER BY COUNT(*) DESC LIMIT 5;", conn))
```

```python
hourly_departures = pd.read_sql("SELECT Start_Station_Id, Start_Station_Name, strftime('%Y-%m-%d-%H:00', Start_Time) AS hour, AVG(Trip_Duration) AS avg_duration_s, COUNT(*) AS departures FROM trips GROUP BY Start_Station_id, strftime('%Y-%m-%d-%H:00', Start_Time);", conn)
```

```python
hourly_departures.shape
```

```python
hourly_departures.to_sql("hourly_departures", conn, if_exists="replace", index=False)
```
