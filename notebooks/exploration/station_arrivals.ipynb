```python
import sqlite3
import pandas as pd
conn = sqlite3.connect("../../data/processed/bikeshare.db")
print(pd.read_sql("SELECT End_Station_Id, COUNT(*) FROM trips WHERE date(End_Time) = '2026-01-01' GROUP BY End_Station_id ORDER BY COUNT(*) DESC;", conn))
```

```python
print(pd.read_sql("SELECT End_Station_Id, COUNT(*) FROM trips WHERE date(End_Time) = '2026-01-01' AND End_Station_Id IS NOT NULL GROUP BY End_Station_id ORDER BY COUNT(*) DESC;", conn))
```

```python
print(pd.read_sql("SELECT End_Station_Id, End_Station_Name, COUNT(*) FROM trips WHERE date(Start_Time) = '2026-05-04' AND End_Station_Id IS NOT NULL GROUP BY Start_Station_id ORDER BY COUNT(*) DESC;", conn))
```

```python
print(pd.read_sql("SELECT End_Station_Id, End_Station_Name, strftime('%Y-%m-%d-%H:00', End_Time), AVG(Trip_Duration), COUNT(*) FROM trips WHERE End_Station_Id IS NOT NULL GROUP BY End_Station_id, strftime('%Y-%m-%d-%H', End_Time) ORDER BY COUNT(*) DESC LIMIT 5;", conn))
```

```python
hourly_arrivals = pd.read_sql("SELECT End_Station_Id, End_Station_Name, strftime('%Y-%m-%d-%H:00', End_Time) AS hour, AVG(Trip_Duration) AS avg_duration_s, COUNT(*) AS arrivals FROM trips WHERE End_Station_Id IS NOT NULL GROUP BY End_Station_id, strftime('%Y-%m-%d-%H:00', End_Time);", conn)
```

```python
hourly_arrivals.shape
```

```python
hourly_arrivals.to_sql("hourly_arrivals", conn, if_exists= "replace", index = False)
conn.close()
```
